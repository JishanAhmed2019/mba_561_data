## Task 4 (4 points) - Make necessary datatype conversions

The `Date` column in `dfi` and the `WeekStarting` column in `dfw` contain dates but initially load as `object`. Both should be converted to pandas datetime values.

```python
# Convert date columns to datetime
dfi["Date"] = pd.to_datetime(dfi["Date"]).dt.normalize()
dfw["WeekStarting"] = pd.to_datetime(dfw["WeekStarting"]).dt.normalize()

# Verify the conversions
print("Updated information for dfi:")
dfi.info()

print("\nUpdated information for dfw:")
dfw.info()
```

```python
# Check that the date conversions really worked

# If any date failed to parse it becomes NaT, so this count should be 0.
print("Bad dates in dfi:", dfi["Date"].isna().sum())
print("Bad dates in dfw:", dfw["WeekStarting"].isna().sum())

# Date math is now possible, which was the whole point.
print("\ndfi date range:", dfi["Date"].min(), "to", dfi["Date"].max())
print("dfw date range:", dfw["WeekStarting"].min(), "to", dfw["WeekStarting"].max())

print("\nTransactions by day of week:")
print(dfi["Date"].dt.day_name().value_counts())

# Every WeekStarting should land on the same weekday.
print("\nWeekStarting always falls on:", dfw["WeekStarting"].dt.day_name().unique())
```

### Why I made these datatype conversions

I converted `dfi["Date"]` and `dfw["WeekStarting"]` from `object` to `datetime64[ns]`. These columns contain dates, so storing them as datetime values is more appropriate than storing them as text. This conversion is also important for later tasks because I need to aggregate sales by date, add days to `WeekStarting`, sort chronologically, and join the two DataFrames using dates. The updated dataframe information confirms that both columns are now stored as datetime values.

I left the other columns unchanged because their existing datatypes are appropriate. `Price` and `TotalDue` are numeric measurements stored as `float64`, while `Quantity` and `Week` are whole numbers stored as `int64`. The daily temperature columns are already numeric (`float64`). Columns such as `StoreNumber`, `TransactionNumber`, and `CustomerCode` are identifiers rather than numerical measurements, so keeping them as text is appropriate. The remaining object columns contain names or categories and do not require conversion for the tasks in this analysis.
